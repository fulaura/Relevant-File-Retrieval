```python
from langchain_community.llms import Ollama
from langchain.prompts import PromptTemplate
from langchain.chains import LLMChain
```

```python
llm = Ollama(model="dolphin-llama3:8b")

# === Step 2: Define a simple prompt template ===
prompt = PromptTemplate(
    input_variables=["question"],
    template="Сұрақ: {question}\nЖауап:"
)

# === Step 3: Wrap in a LangChain LLMChain ===
llm_chain = LLMChain(llm=llm, prompt=prompt)

# === Step 4: Run the chain ===
question = "Қазақстан Республикасының мемлекеттік тілі қандай?"
response = llm_chain.run(question)

print("Жауап:", response)
```
